```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
url = 'https://oto.com.vn/mua-ban-xe/p'
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
}

car_links = []
for p in range(1,150):
    # Lấy nội dung HTML
    response = requests.get(url+str(p), headers=headers)
    soup = BeautifulSoup(response.text, 'html.parser')

    # Tìm tất cả thẻ <a> có đường dẫn bài đăng
    links = soup.find_all('a', class_='rt pdt-per-75')

    # Lọc chỉ những link thuộc phần /mua-ban-xe-
    for a in links:
        href = a['href']
        if href.startswith('/mua-ban-xe-'):
            full_link = f"https://oto.com.vn{href}"
            car_links.append(full_link)


for link in car_links:
    print(link)

print(f"\nTổng cộng: {len(car_links)} link.")
df = pd.DataFrame({'Link': car_links})
df.to_csv('car_links.csv', index=False, encoding='utf-8-sig')
```
