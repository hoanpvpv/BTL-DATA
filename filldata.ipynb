```python
import pandas as pd
import numpy as np

data = pd.read_csv("heart_2022_with_nans.csv")
data
```

```python
numberCols = data.select_dtypes('float').columns
print(numberCols)
print(data[numberCols].isna().sum())
```

```python
data[numberCols] = data[numberCols].fillna(data[numberCols].mean())
print(data[numberCols].isna().sum())
```

```python
textCols = data.select_dtypes(include='object').columns
print(textCols)

print(data[textCols].isna().sum())
```

```python
for col in textCols:
    print(col,data[col].value_counts().idxmax(), data[col].value_counts().max())
    

```

```python
data[textCols]=data[textCols].fillna(data[textCols].value_counts().max())
print(data[textCols].isna().sum())
```

```python

for col in numberCols:
    dt = sorted(data[col])
    Q1 = np.percentile(dt,25)
    Q2 = np.median(dt)
    Q3 = np.percentile(dt,75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5*IQR
    upper_bound = Q3 +1.5*IQR
    outliers = [i for i in dt if i < lower_bound or i > upper_bound]
    print(f"{col}: {outliers}")
    
    dataNoOutLine = data[(data[col] >= lower_bound) & (data[col] <= upper_bound)]

dataNoOutLine.shape
    
```
