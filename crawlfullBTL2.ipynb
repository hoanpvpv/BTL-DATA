```python
import pandas as pd
from bs4 import BeautifulSoup
import requests


df = pd.read_csv('car_links.csv')
links = df['Link'].tolist()

headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
}

cars = []

for link in links:
    try:
        response = requests.get(link, headers=headers)
        response.encoding = response.apparent_encoding
        soup = BeautifulSoup(response.text, 'html.parser')

        # Các trường cơ bản
        createAt = soup.find('span', class_='date').get_text(strip=True) if soup.find('span', class_='date') else ''
        name = soup.find('h1', class_='title-detail').get_text(strip=True) if soup.find('h1', class_='title-detail') else ''
        price = soup.find('span', class_='price').get_text(strip=True) if soup.find('span', class_='price') else ''

        # Các trường trong danh sách thông tin
        def get_field(label):
            tag = soup.find('label', text=label)
            if tag:
                li = tag.find_parent('li')
                if li:
                    return li.get_text(strip=True).replace(label, '').strip()
            return ''

        namSX = get_field('Năm SX:')
        xuatXu = get_field('Xuất xứ:')
        address = get_field('Tỉnh thành:')
        shape = get_field('Kiểu dáng:')
        soKMdadi = get_field('Km đã đi:')
        hoSo = get_field('Hộp số:')
        status = get_field('Tình trạng:')
        nhienLieu = get_field('Nhiên liệu:')

        cars.append([createAt, name, namSX, xuatXu, address, shape, soKMdadi, hoSo, status, nhienLieu, price])

        print(f"Đã lấy: {[createAt, name, namSX, xuatXu, address, shape, soKMdadi, hoSo, status, nhienLieu, price]}")
    except Exception as e:
        print(f"Lỗi ở link {link}: {e}")

# Lưu ra CSV
df_out = pd.DataFrame(cars, columns=["Ngày đăng bài", "Tên", "Năm sản xuất", "Xuất xứ", "Địa điểm", "Kiểu dáng", "Số km đã đi", "Hộp số", "Trạng thái", "Nhiên Liệu", "Giá"])
df_out.to_csv('oto_full_data.csv', index=False, encoding='utf-8-sig')
print("\n Đã lưu dữ liệu vào oto_full_data.csv")
```
